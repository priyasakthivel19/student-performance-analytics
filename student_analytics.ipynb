```python
import pandas as pd
import numpy as np
print("All good!")
```

```python
df = pd.read_csv("student_performance_dataset.csv")
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
df['Python'] = df['Python'].fillna(df['Python'].mean())
df['SQL'] = df['SQL'].fillna(df['SQL'].mean())

# Verify pannunga
df.isnull().sum()
```

```python
df['Name'] = df['Name'].str.strip()
print("Names cleaned!")
```

```python
df['Attendance'] = df['Attendance'].apply(lambda x: x/10 if x > 100 else x)
df['Attendance'].describe()
```

```python
df['Aptitude'] = df['Aptitude'].apply(lambda x: abs(x) if x < 0 else x)
df['Aptitude'].describe()
```

```python
print("Before removing duplicates:", df.shape)
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python
df.isnull().sum()
df.describe()
```

```python
df.to_csv("cleaned_students.csv", index=False)
print("Cleaned file saved!")
```

```python
subject_avg = df[['Python','SQL','Maths','DBMS']].mean().sort_values()
print(subject_avg)
```

```python
correlation = df['Attendance'].corr(df['Python'])
print("Attendance vs Python correlation:", correlation)
```

```python
skills = ['PowerBI_Skill','Excel_Skill','HTML_CSS_Skill','Java_Skill']
for skill in skills:
    yes_percent = (df[skill] == 'Yes').mean() * 100
    print(f"{skill}: {yes_percent:.1f}% students have this skill")
```

```python
def recommend_career(row):
    if row['Python'] >= 75 and row['SQL'] >= 70:
        return "Data Analyst"
    elif row['DBMS'] >= 75 and row['Aptitude'] >= 75:
        return "Software Developer"
    elif row['HTML_CSS_Skill'] == 'Yes' and row['Java_Skill'] == 'No':
        return "UI/UX / Frontend"
    elif row['Java_Skill'] == 'Yes' and row['DBMS'] >= 65:
        return "Software Developer"
    else:
        return "Needs Skill Improvement"

df['Recommended_Career'] = df.apply(recommend_career, axis=1)

# Career distribution paarunga
print(df['Recommended_Career'].value_counts())
```

```python
df.to_csv("cleaned_students.csv", index=False)
print("Final file saved with career recommendations!")
```
